# Imports

```python
from pathlib import Path

import json
import ollama
```

# Params

```python
PROJECT_ROOT = Path.cwd().parent

OUTPUT_DIR = PROJECT_ROOT / "data" / "raw"
```

```python
MODEL = "qwen3:4b"
```

```python
USER_TOPIC = """
I want to learn about multi-agent systems applied to data analytics.
"""
```

# Prompt and Response

```python
prompt = f"""
You are an expert AI researcher.

Your task is to generate an optimized search query for arXiv.

The query should:

- maximize relevant research papers
- minimize irrelevant papers
- use AND and OR when appropriate
- be concise
- use English keywords only

Return ONLY the search query.

User topic:

{USER_TOPIC}
"""
```

```python
response = ollama.chat(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ],
)
```

```python
query = response["message"]["content"].strip()

print(query)
```

# Save processed Query

```python
config = {
    "topic": USER_TOPIC.strip(),
    "query": query,
}

with open(
    OUTPUT_DIR / "arxiv_query.json",
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        config,
        file,
        indent=4,
        ensure_ascii=False,
    )
```

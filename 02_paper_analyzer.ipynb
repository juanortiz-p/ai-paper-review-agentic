# Imports

```python
from pathlib import Path

import ollama
import pandas as pd
```

# Paths

```python
PROJECT_ROOT = Path.cwd().parent

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "arxiv_search.csv"
)

df = pd.read_csv(INPUT_PATH)
```

```python
paper = df.iloc[0]
```

```python
MODEL = "qwen3:4b"
```

# Prompting

```python
prompt = f"""
You are an experienced AI researcher.

Read the following research paper abstract.

Paper title:
{paper["title"]}

Abstract:
{paper["summary"]}

Explain:

1. What problem does the paper solve?
2. What is the main contribution?
3. Why is it interesting?
4. What kind of applications could use this work?

Keep the answer concise.
"""
```

# Get response from local model

```python
response = ollama.chat(
    model=MODEL,
    messages=[
        {
            "role": "user",
            "content": prompt,
        }
    ],
)
```

```python
print(response["message"]["content"])
```
